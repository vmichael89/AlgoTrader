```python
# Clean and structure the extracted data
data = """
1.0290 (€1.2bn)
1.0300 (€3.1bn)
1.0325 (€5.6bn)
1.0350 (€4.8bn)
1.0390 (€1.7bn)
1.0400 (€6.3bn)
1.0425 (€4.1bn)
1.0430 (€1.8bn)
1.0440 (€3.2bn)
1.0450 (€8.3bn)
1.0490 (€860m)
1.0500 (€10.3bn)
1.0510 (€768m)
1.0525 (€6.6bn)
1.0550 (€3.7bn)
1.0555 (€1.4bn)
1.0570 (€615m)
1.0575 (€1.3bn)
1.0580 (€1.5bn)
1.0590 (€1.0bn)
1.0600 (€6.4bn)
"""

# Parse the data into a DataFrame
rows = []
for line in data.strip().split("\n"):
    strike, value = line.split(" (€")
    strike = float(strike)
    value = value.replace("bn)", "").replace("m)", "")
    multiplier = 1_000_000_000 if "bn" in line else 1_000_000
    value = float(value) * multiplier
    rows.append((strike, value))

df = pd.DataFrame(rows, columns=["Strike Price", "Value"])

plt.figure(figsize=(10, 6))
plt.barh(
    df["Strike Price"], 
    df["Value"]/1e9, 
    color='skyblue', 
    edgecolor='black', 
    height=0.0005  # Thinner bars for spacing
)
plt.xlabel("Value (billion €)", fontsize=12)
plt.ylabel("Strike Price", fontsize=12)
plt.title("FX Options Expiry Values by Strike Price\nEURUSD | 20/12/2024 | 07:02 GMT+1", fontsize=14)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()

# Display the updated chart
plt.show()
```
