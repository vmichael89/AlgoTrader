## News Data

Download news events from forex factory

```python
scraper = cloudscraper.create_scraper()
url = r"https://www.forexfactory.com/calendar?week=dec01.2024"

response = scraper.get(url)

events_matches = re.findall(r'"events":\[(\{[^\]]*\})\]', response.text)
parsed_data = json.loads("[" + ",".join(events_matches) + "]")

df_news = pd.DataFrame(parsed_data)

utc_correction = datetime.datetime.now() - datetime.datetime.utcnow()

df_news["datetime"] = df_news["dateline"].map(datetime.datetime.fromtimestamp) - utc_correction

# Delete some columns
df_news = df_news[["datetime", "currency", "name", "impactTitle"]]
df_news.tail(15)
```

## Tick Data

For each news event, fetch **bid and ask tick data** from 10 minutes before until 20 minutes after the event.

Select the major currency pairs. For US news, select EURUSD, GBPUSD and USDJPY.


```python
df_sym = pd.DataFrame([sym._asdict() for sym in mt.symbols_get()])

# Get all available symbols related to Forex
df_sym_red = df_sym[["currency_base", "currency_profit", "name", "path"]]
df_sym_forex = df_sym_red[df_sym_red.path.str.startswith("Forex")]
df_sym_forex
print(*df_sym_red.name[:], sep="\n")
```

```python
df_news_red = df_news[df_news["impactTitle"].str.contains("High")]
df_news_red = df_news.sort_values(by='impactTitle')
df_news_red
```

```python
for row, test_news in df_news_red.iterrows():
    # print(row)

    # Related pairs
    df_pair = df_sym_forex[df_sym_forex.name.str.contains(test_news.currency)]
    df_pair = df_pair[df_pair.name.str.contains("USD")]

    if len(df_pair) == 0:
        continue

    symbol = df_pair.name.iloc[0]
    # print(symbol)

    start_time = (test_news.datetime - pd.Timedelta(minutes=1) + pd.Timedelta(hours=3)).to_pydatetime()
    end_time = (test_news.datetime + pd.Timedelta(minutes=1) + pd.Timedelta(hours=3)).to_pydatetime()
    # print(start_time, end_time)

    ticks = mt.copy_ticks_range(symbol, start_time, end_time, mt.COPY_TICKS_ALL)
    df_ticks = pd.DataFrame(ticks)
    df_ticks['datetime'] = pd.to_datetime(df_ticks['time_msc'], unit='ms')
    df_ticks.set_index('datetime', inplace=True)

    df_ticks['spread'] = (df_ticks['ask'] - df_ticks['bid'])

    tickvolume_seconds = 1
    tickvolume = df_ticks['bid'].resample(f'{tickvolume_seconds}S').count()


    # Calculate the range of price (bid/ask)
    price_min, price_max = df_ticks[['bid', 'ask']].min().min(), df_ticks[['bid', 'ask']].max().max()
    mid_price = price_max / 2 + price_min / 2
    price_range = price_max - price_min
    y_range = [price_min - 0.1 * price_range, price_max + 0.1 * price_range]

    # Calculate the range of spread
    spread_min, spread_max = df_ticks['spread'].min(), df_ticks['spread'].max()
    mid_spread = spread_max / 2 + spread_min / 2
    y3_range = [spread_min, spread_max]

    fig = go.Figure()

    fig.add_trace(go.Bar(x=tickvolume.index, y=tickvolume, name='Tick Volume', marker_color='orange', opacity=0.4, yaxis='y2'))
    fig.add_trace(go.Scatter(x=df_ticks.index, y=df_ticks['spread'], name='Spread', fillcolor='rgba(80, 80, 80, 0.2)', fill='tozeroy', mode='none', yaxis='y3'))
    fig.add_trace(go.Scatter(x=df_ticks.index, y=df_ticks['bid'], name='Bid', line_color='blue', line_width=.5, mode='lines+markers', marker_symbol='cross', marker_size=2))
    fig.add_trace(go.Scatter(x=df_ticks.index, y=df_ticks['ask'], name='Ask', line_color='green', line_width=.5, mode='lines+markers',marker_symbol='cross', marker_size=2))

    # Update layout to add a secondary y-axis
    fig.update_layout(
        title=f"{symbol} - {test_news.datetime}<br>{test_news.impactTitle} - {test_news['name']}<br>Bid, Ask, and Tick Volume ({tickvolume_seconds}-Second Bars)",
        xaxis=dict(
            title='Time', 
            domain=[0.05, 0.9]),
        yaxis=dict(
            title='Price', 
            side='left', showgrid=False, 
            dtick=0.0001, range=y_range, domain=[0, 0.9]),
        yaxis2=dict(
            title='Tick Volume', 
            side='right', showgrid=False, 
            anchor='x', position=0.9, range=[0, 100], domain=[0, 0.4]),
        yaxis3=dict(
            title='Spread', 
            side='right', showgrid=False, 
            anchor='x', position=0.9, range=y3_range[::-1], domain=[0.6, 0.9]),
        plot_bgcolor='rgba(0,0,0,0)',
        template='plotly_white',
        dragmode='zoom',
        hovermode='x'
    )

    fig.show()
```
